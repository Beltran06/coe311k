```python
import numpy as np
import math
import matplotlib.pyplot as plt

def sin_taylor(x, n_terms):
    
    approximation = 0.0
    for n in range(n_terms):
        term = (-1)**n * x**(2*n + 1) / math.factorial(2*n + 1)
        approximation += term
    return approximation
```

```python
x_values = np.arange(0, 10.1, 0.1)
exact_sin = np.sin(x_values)

terms = [1, 2, 3, 4]
approximations = {}
errors = {}

for n in terms:
    approximations[n] = np.array([sin_taylor(x, n) for x in x_values])
    errors[n] = np.abs(approximations[n] - exact_sin)
```

```python
plt.figure()
for n in terms:
    plt.plot(x_values, errors[n], label=f"{n} term(s)")

plt.xlabel("x")
plt.ylabel("Absolute Error")
plt.title("Truncation Error vs x for sin(x) Taylor Approximation")
plt.legend()
plt.grid(True)
plt.show()

x_test_values = [1, 5, 10]

for x_test in x_test_values:
    exact = np.sin(x_test)
    error_list = []

    for n in terms:
        approx = sin_taylor(x_test, n)
        error_list.append(abs(approx - exact))

    plt.figure()
    plt.plot(terms, error_list, marker='o')
    plt.xlabel("Number of Terms")
    plt.ylabel("Absolute Error")
    plt.title(f"Error vs Number of Terms at x = {x_test}")
    plt.grid(True)
    plt.show()
```

As additional terms are included in the Taylor series expansion, the truncation error decreases. However, the decrease in error is not uniform across all values of x. For small values of x near zero, even a small number of terms provides an accurate approximation. As x increases, the truncation error becomes significantly larger for the same number of terms, since higher powers of x grow rapidly. Including more terms improves accuracy for larger x, but convergence is slower compared to small x. The Taylor series approximation is most effective near the expansion point and requires many more terms to remain accurate for larger values of x.
