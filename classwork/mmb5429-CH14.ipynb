```python
import numpy as np
import matplotlib.pyplot as plt

# Setup Data
a, b = 0, 2 * np.pi
N = 100
x = np.linspace(a, b, N)
h = (b - a) / (N - 1)

def f(x):
    return np.sin(x)

# Derivative Approximations
# Forward Difference: (f(x + h) - f(x)) / h
df_forward = (f(x + h) - f(x)) / h

# Backward Difference: (f(x) - f(x - h)) / h
df_backward = (f(x) - f(x - h)) / h

# Central Difference: (f(x + h) - f(x - h)) / (2 * h)
df_central = (f(x + h) - f(x - h)) / (2 * h)

# True Derivative for comparison
df_true = np.cos(x)

# Plotting (4 Subplots)
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# Subplot 1: True Derivative
axs[0, 0].plot(x, df_true, color='black', label='cos(x)')
axs[0, 0].set_title('True Derivative: $\cos(x)$')
axs[0, 0].grid(True)
axs[0, 0].legend()

# Subplot 2: Forward Difference
axs[0, 1].plot(x, df_true, 'k--', alpha=0.3, label='True')
axs[0, 1].plot(x, df_forward, color='red', label='Forward')
axs[0, 1].set_title('Forward Difference')
axs[0, 1].grid(True)
axs[0, 1].legend()

# Subplot 3: Backward Difference
axs[1, 0].plot(x, df_true, 'k--', alpha=0.3, label='True')
axs[1, 0].plot(x, df_backward, color='green', label='Backward')
axs[1, 0].set_title('Backward Difference')
axs[1, 0].grid(True)
axs[1, 0].legend()

# Subplot 4: Central Difference
axs[1, 1].plot(x, df_true, 'k--', alpha=0.3, label='True')
axs[1, 1].plot(x, df_central, color='blue', label='Central')
axs[1, 1].set_title('Central Difference')
axs[1, 1].grid(True)
axs[1, 1].legend()

# labels
for ax in axs.flat:
    ax.set_xlabel('x')
    ax.set_ylabel("f'(x)")

plt.tight_layout()
plt.show()
```
