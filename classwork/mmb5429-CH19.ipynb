```python
import numpy as np
import matplotlib.pyplot as plt

# Inputs
t_start = 0
t_end = 30
y0 = 0
step_sizes = [0.1, 0.01, 0.001, 0.00001, 0.0000001]

# Define the differential equation and analytical solution
def f(t, y):
    return -1000*y + 3000 - 2000*np.exp(-t)

def analytical_sol(t):
    return 3 - 0.998*np.exp(-1000*t) - 2.002*np.exp(-t)

plt.figure(figsize=(10, 6))

# Numerical Integration Loop
for h in step_sizes:
    # Explicitly calculating number of steps
    num_steps = int((t_end - t_start) / h)
    
    t_values = np.zeros(num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    
    t_values[0] = t_start
    y_values[0] = y0
    
    for i in range(num_steps):
        # Euler's Formula: y_new = y_old + h * f(t_old, y_old)
        y_values[i+1] = y_values[i] + h * f(t_values[i], y_values[i])
        t_values[i+1] = t_values[i] + h
        
        # Emergency break for unstable h (0.1 and 0.01)
        if abs(y_values[i+1]) > 1e4:
            t_values = t_values[:i+2]
            y_values = y_values[:i+2]
            break
            
    plt.plot(t_values, y_values, label=f'h = {h}')

# Plot comparison
t_exact = np.linspace(t_start, t_end, 500)
plt.plot(t_exact, analytical_sol(t_exact), 'k--', label='Analytical')

plt.ylim(-2, 5) # Limit y-axis to actually see the stable results
plt.xlabel('Time (t)')
plt.ylabel('y(t)')
plt.title('Forward Euler Stability Challenge')
plt.legend()
plt.grid(True)
plt.show()
```
