```python
import numpy as np

A = np.array([
[0.582745, 0.48 , 0.10 , 0.  , 0.  ],
[0.48 , 1.044129, 0.46 , 0.10 , 0.  ],
[0.10 , 0.46 , 1.10431 , 0.44 , 0.10 ],
[0.  , 0.10 , 0.44 , 0.963889, 0.42 ],
[0.  , 0.  , 0.10 , 0.42 , 0.522565]], dtype=float)

b = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)

# Initial guess
x0 = np.zeros(len(b))

# Tolerance and max iterations
tol = 1e-6
max_iterations = 100


import numpy as np

def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations


def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    
    for k in range(max_iterations):
        x_new = x.copy()
        
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        
        x = x_new
        
    return x, max_iterations



# Solve using Jacobi
x_jacobi, iter_jacobi = jacobi(A, b, x0, tol, max_iterations)

# Solve using Gauss-Seidel
x_gs, iter_gs = gauss_seidel(A, b, x0, tol, max_iterations)

print("Jacobi Solution:")
print(x_jacobi)
print("Iterations:", iter_jacobi)

print("\nGauss-Seidel Solution:")
print(x_gs)
print("Iterations:", iter_gs)
```

Gauss-Seidel successfully converged to the solution (all 1s) in 20 iterations, while Jacobi failed to reach the tolerance with the 100 iteration limit.
