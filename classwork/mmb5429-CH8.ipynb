```python
def is_diagonally_dominant(matrix):
    n = len(matrix)
    for i in range(n):
        # absolute value of the diagonal element
        diagonal_element = abs(matrix[i][i])
        
        # sum of the absolute values of the other elements in the row
        row_sum = sum(abs(matrix[i][j]) for j in range(n) if i != j)
        
        if diagonal_element <= row_sum:
            return False
            
    return True

def display_matrix(matrix):
    for row in matrix:
        print(row)

# Example Usage:
matrix_a = [[4, 2, 1], 
            [1, 5, 2], 
            [0, 1, 3]]

result = is_diagonally_dominant(matrix_a)

print(f"Is the matrix diagonally dominant? {result}")
display_matrix(matrix_a)
```
