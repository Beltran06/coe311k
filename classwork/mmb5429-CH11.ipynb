```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline

# Define known data points
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 0, 1, 0, 1])

# Cubic Spline Interpolation
cs = CubicSpline(x, y, bc_type='natural')

# Polyfit
poly_coeffs = np.polyfit(x, y, 5)
poly_func = np.poly1d(poly_coeffs)

# Generate new x values for comparison
x_interp = np.linspace(0, 5, 100)
y_spline = cs(x_interp)
y_poly = poly_func(x_interp)

# Visualization for comparison
plt.plot(x, y, 'ko', label='Original Data')
plt.plot(x_interp, y_spline, label='CubicSpline')
plt.plot(x_interp, y_poly, '--', label='Polyfit (deg 5)')
plt.legend()
plt.show()
```
